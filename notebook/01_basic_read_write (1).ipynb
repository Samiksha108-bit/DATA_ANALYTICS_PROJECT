## 1. Reading a file with plain Python

```python
# Read the whole file as text
with open("hospital_data.csv", "r") as f:
    content = f.read()

print("Total characters:", len(content))
print(content[:300])
```

```python
# Read line by line
with open("hospital_data.csv", "r") as f:
    lines = f.readlines()

print("Total lines (including header):", len(lines))
print("Header:", lines[0].strip())
print("First record:", lines[1].strip())
```

## 2. Writing a file with plain Python

```python
# Write a small text file
with open("notes.txt", "w") as f:
    f.write("Hospital data project\n")
    f.write("Step 1: read and write files\n")

# Append to it
with open("notes.txt", "a") as f:
    f.write("Step 2: pandas and numpy\n")

# Read it back
with open("notes.txt", "r") as f:
    print(f.read())
```

## 3. The `csv` module

```python
import csv

with open("hospital_data.csv", "r", newline="") as f:
    reader = csv.reader(f)
    header = next(reader)
    rows = list(reader)

print("Columns:", header)
print("Number of records:", len(rows))
print("First 3 records:")
for r in rows[:3]:
    print(r)
```

```python
# DictReader gives each row as a dictionary
with open("hospital_data.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    records = list(reader)

print(records[0])
print("Department of first patient:", records[0]["Department"])
```

```python
# Write a subset of rows (Emergency admissions only) to a new CSV
emergency = [r for r in records if r["Admission_Type"] == "Emergency"]

with open("emergency_only.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=header)
    writer.writeheader()
    writer.writerows(emergency)

print("Saved", len(emergency), "emergency records to emergency_only.csv")
```

## 4. Reading and writing with pandas

```python
import pandas as pd

df = pd.read_csv("hospital_data.csv")
df.head()
```

```python
# Write to different formats
df.head(20).to_csv("sample.csv", index=False)       # CSV

print("Files saved: sample.csv")
```

```python
# Read them back
print(pd.read_csv("sample.csv").shape)
```
