```python
import pandas as pd
import numpy as np

df = pd.read_csv("hospital_data.csv")
print("Shape:", df.shape)
df.head()
```

## Part A: Pandas

### 1. Understand the data

```python
df.info()
```

```python
df.describe()
```

```python
print("Columns:", df.columns.tolist())
print()
print(df.dtypes)
```

### 2. Selecting data

```python
# One column / multiple columns
print(df["Age"].head())
df[["Patient_ID", "Department", "Length_of_Stay_Days"]].head()
```

```python
# loc (by label/condition) and iloc (by position)
print(df.loc[0, "Department"])
print(df.iloc[0:3, 0:4])
```

### 3. Filtering rows

```python
# Emergency patients
emergency = df[df["Admission_Type"] == "Emergency"]
print("Emergency admissions:", len(emergency))

# Multiple conditions: Cardiology patients with long stays
cardio_long = df[(df["Department"] == "Cardiology") & (df["Length_of_Stay_Days"] > 7)]
print("Cardiology with stay > 7 days:", len(cardio_long))
cardio_long.head()
```

### 4. Missing values and duplicates

```python
print("Missing values per column:")
print(df.isnull().sum())
print()
print("Duplicate rows:", df.duplicated().sum())
```

### 5. Sorting, counting and grouping

```python
# Sort by length of stay
df.sort_values("Length_of_Stay_Days", ascending=False).head()
```

```python
print(df["Department"].value_counts())
print()
print(df["Gender"].value_counts())
```

```python
# Average stay and waiting time per department
df.groupby("Department")[["Length_of_Stay_Days", "Waiting_Time_Mins"]].mean().round(2)
```

```python
# Multiple aggregations
df.groupby("Admission_Type")["Length_of_Stay_Days"].agg(["count", "mean", "min", "max"])
```

### 6. Creating new columns

```python
df["Stay_Category"] = np.where(df["Length_of_Stay_Days"] > 7, "Long", "Short")
df[["Patient_ID", "Length_of_Stay_Days", "Stay_Category"]].head()
```

```python
# Convert the date column and extract parts
df["Admission_Date"] = pd.to_datetime(df["Admission_Date"])
df["Admission_Year"] = df["Admission_Date"].dt.year
df["Admission_Month"] = df["Admission_Date"].dt.month_name()
df[["Admission_Date", "Admission_Year", "Admission_Month"]].head()
```

## Part B: NumPy

### 1. Arrays from DataFrame columns

```python
stay = df["Length_of_Stay_Days"].to_numpy()
print(type(stay), stay.shape, stay.dtype)
print(stay[:10])
```

### 2. Basic statistics

```python
print("Mean   :", np.mean(stay))
print("Median :", np.median(stay))
print("Std dev:", np.std(stay))
print("Min/Max:", stay.min(), stay.max())
print("25th/75th percentile:", np.percentile(stay, [25, 75]))
```

### 3. Handling NaN with NumPy

```python
age = df["Age"].to_numpy()
print("Number of NaN in Age:", np.isnan(age).sum())
print("Mean (ignoring NaN):", np.nanmean(age))
print("Median (ignoring NaN):", np.nanmedian(age))
```

### 4. Array operations and conditions

```python
wait = df["Waiting_Time_Mins"].to_numpy()
wait_hours = wait / 60                      # vectorised math, no loop
print(np.round(wait_hours[:5], 2))

# Boolean masks
print("Patients waiting more than 2 hours:", np.sum(wait > 120))
```

```python
# np.where: label blood pressure levels
bp = df["Systolic_BP"].to_numpy()
bp_level = np.where(bp >= 140, "High", np.where(bp >= 120, "Elevated", "Normal"))
print(pd.Series(bp_level).value_counts())
```

### 5. Correlation

```python
numeric_df = df[["Age", "Num_Comorbidities", "Systolic_BP", "Waiting_Time_Mins", "Length_of_Stay_Days"]]
numeric_df.corr().round(2)
```

```python
# NumPy correlation between two columns (drop NaN rows first)
pair = df[["Num_Comorbidities", "Length_of_Stay_Days"]].dropna()
print(np.corrcoef(pair["Num_Comorbidities"], pair["Length_of_Stay_Days"])[0, 1])
```
